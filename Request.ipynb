```python
import requests

url = "https://listado.mercadolibre.com.pe/laptop"

# Añadimos un User-Agent para simular un navegador real
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}

try:
    # Pasamos los headers en la petición
    respuesta = requests.get(url, headers=headers)
    respuesta.raise_for_status()
    print(f"¡Conexión exitosa! Estado: {respuesta.status_code}")
    
except requests.exceptions.RequestException as e:
    # Capturamos cualquier error de conexión
    print(f"Error al intentar conectarse: {e}")
```
